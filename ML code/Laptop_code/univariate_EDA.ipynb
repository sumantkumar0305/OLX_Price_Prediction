## Statistical Summary

The `Price` variable contains **3,859 observations** with **no missing values** and **410 unique values**.

| Metric | Original Scale | Log-Transformed Scale |
|---|---:|---:|

| **Skewness** | **4.25** | **0.09** |
| **Kurtosis** | **27.42** | **0.24** |

## Observations and Insights

### 1. Distribution Shape

- The original `Price` variable is **highly right-skewed**, with a skewness of **4.25**.
- The mean (**₹32,755.47**) is considerably higher than the median (**₹20,997**), indicating that high-priced products are pulling the distribution toward the right.
- The **75th percentile is ₹39,000**, while the maximum price reaches **₹4,80,000**, showing a large spread in the upper tail.
- The kurtosis of **27.42** indicates **heavy tails and extreme values** in the original price distribution.

### 2. Outlier Analysis

Using the standard **1.5 × IQR rule**:

- Q1 = ₹12,000
- Q3 = ₹39,000
- IQR = ₹27,000
- Upper Bound = ₹39,000 + (1.5 × ₹27,000)
- **Upper Bound = ₹79,500**

There are approximately **258 observations (6.69%)** above this upper bound.

These observations should not necessarily be removed because they may represent **genuine premium or high-value products** rather than incorrect data.

### 3. Effect of Log Transformation

A natural logarithmic transformation was applied to the `Price` variable.

The transformation produced a significant improvement in the distribution:

- **Skewness:** `4.25 → 0.09`
- **Kurtosis:** `27.42 → 0.24`

This indicates that the log-transformed price distribution is **much closer to a symmetric and approximately normal distribution**.

The transformation also reduces the influence of extremely expensive products and makes the overall variance more stable.

## Conclusion

The `Price` variable is highly skewed and contains several extreme high-value observations. However, these observations may represent valid market prices and should not be removed without further investigation.

The **log transformation is effective in reducing skewness and the influence of extreme values**. Therefore, using a transformed target such as:

````python
def fixProcessor(row):

    lowest_price = row['Price'] - 5000
    upper_bound = row['Price'] + 10000

    temp = data[
        (data['Price'] >= lowest_price) &
        (data['Price'] <= upper_bound) &
        (data['Brand'] == row['Brand'])
    ]['processor'].value_counts()

    if len(temp) > 0:
        row['processor'] = temp.index[0]

    return row
````
