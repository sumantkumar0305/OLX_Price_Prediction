# Bivariate EDA Summary: `Brand` vs `Network`

## 1. Distribution of Network Types Across Brands

- The crosstab analysis shows the number and percentage of **4G and 5G devices for each brand**.
- Some brands, such as **Google, iQOO, and Nothing**, have **100% 5G devices** in this dataset.
- **Itel** is different from most other brands, with approximately **93.75% of its devices using 4G**.
- **Apple** has approximately **81.73% 5G devices**, while the remaining devices use 4G.
- **Samsung** has approximately **81.51% 5G devices**, showing that most Samsung devices in the dataset support 5G.

## 2. Visualization Insights

- The grouped countplot clearly shows that **5G devices are more common than 4G devices for most brands**.
- **Apple and Samsung** have the highest number of devices in the dataset.
- For both Apple and Samsung, the number of **5G devices is much higher than 4G devices**.
- **Itel and Lava** are among the brands where **4G devices are more common than 5G devices**.
- Overall, the visualization shows a clear shift in the smartphone market toward **5G technology**.

## 3. Chi-Square Test

A Chi-Square test was performed to determine whether there is a statistically significant relationship between `Brand` and `Network`.

- **Chi-Square Statistic:** approximately **266.36**
- **Degrees of Freedom:** **16**
- **p-value:** approximately **2.249 × 10⁻⁴⁷**

Since the **p-value is much smaller than 0.05**, we reject the **null hypothesis**.

### Null Hypothesis (H₀)

There is **no significant relationship** between `Brand` and `Network`.

### Alternative Hypothesis (H₁)

There **is a significant relationship** between `Brand` and `Network`.

## 4. Conclusion

- The Chi-Square test shows that there is a **statistically significant relationship between Brand and Network type**.
- This means the distribution of **4G and 5G devices is different across brands**.
- Most brands in the dataset have a larger number of **5G devices**.
- Some brands, such as **Itel**, are more concentrated in the **4G category**.
- Therefore, based on this dataset, a phone's **network type is associated with its brand**.

# Bivariate EDA Summary: `Price` vs `Accessories_Included`

## 1. Price Distribution Based on Accessories

- The grouped analysis shows how the average phone price changes based on the number of accessories included.
- Phones with **0 accessories** are the most common group, with **1,665 devices**.
- The average price of phones with **0 accessories** is approximately **₹27,464**.
- Phones with **1 to 4 accessories** have average prices ranging from approximately **₹30,769 to ₹36,842**.
- Phones with **5 accessories** have a higher average price of approximately **₹55,305**, but this group contains only **17 devices**.
- Phones with **6 accessories** have an average price of **₹1,20,000**, but there is only **1 device** in this category.
- Therefore, the result for 6 accessories should not be considered representative because the sample size is too small.

## 2. Boxplot and Violin Plot Analysis

- The boxplot and violin plot show the distribution of phone prices for different numbers of accessories.
- For accessory categories from **0 to 4**, most phones are concentrated in the **lower and middle price range**.
- These groups also show a **right-skewed distribution**, with a small number of expensive phones.
- The **0 accessories** category contains many high-price outliers.
- This may happen because some premium phones are sold without additional accessories such as chargers or earphones.

## 3. Average Price Bar Chart

- The bar chart compares the **average price** of phones across different accessory categories.
- The average price generally increases for some categories with more accessories.
- However, categories with very few observations, such as **6 accessories**, can show unusually high average prices.
- Therefore, both the **mean price and the number of devices** should be considered when interpreting the results.

## 4. ANOVA Test

A One-Way ANOVA test was performed to determine whether the average price is significantly different across accessory groups.

- **F-statistic:** approximately **7.34**
- **p-value:** approximately **9.59 × 10⁻⁹**

Since the **p-value is much smaller than 0.05**, we reject the **null hypothesis**.

### Null Hypothesis (H₀)

The average price is the same for all accessory groups.

### Alternative Hypothesis (H₁)

At least one accessory group has a significantly different average price.

## 5. Conclusion

- There is a **statistically significant relationship between `Accessories_Included` and `Price`**.
- This means the average price differs across different accessory groups.
- However, this does **not necessarily mean that including more accessories directly causes a higher price**.
- Some groups have very few observations, so their average prices should be interpreted carefully.
- Further analysis can help determine which specific accessory groups have significantly different prices.

# Bivariate EDA Summary: `Price` vs `Warranty`

## 1. Price Statistics Based on Warranty

- The grouped analysis shows a clear difference in phone prices based on warranty status.

### Phones Without Warranty

- **2,175 devices** do not have a warranty.
- The average price is approximately **₹28,257**.
- The median price is approximately **₹19,500**.

### Phones With Warranty

- **145 devices** have a warranty.
- The average price is approximately **₹46,741**.
- The median price is approximately **₹35,000**.

- Therefore, phones with a warranty have a **higher average and median price** than phones without a warranty.

## 2. Boxplot and Violin Plot Analysis

- The boxplot and violin plot show the price distribution for phones with and without a warranty.
- Phones without a warranty are mostly concentrated in the **budget to mid-range price segment**.
- However, the `No` warranty category also contains several high-price outliers.
- The maximum price in the `No` category reaches approximately **₹229,900**.
- Phones with a warranty generally have a **higher price distribution**, indicating that warranties are more common among relatively expensive phones.

## 3. Average Price Bar Chart

- The bar chart clearly shows that phones with a warranty have a much higher average price.
- The average price of phones with a warranty is significantly higher than the average price of phones without a warranty.

## 4. ANOVA Test

A One-Way ANOVA test was performed to determine whether the difference in average price between the two warranty groups is statistically significant.

- **F-statistic:** approximately **57.52**
- **p-value:** approximately **4.81 × 10⁻¹⁴**

Since the **p-value is much smaller than 0.05**, we reject the **null hypothesis**.

### Null Hypothesis (H₀)

The average price is the same for phones with and without a warranty.

### Alternative Hypothesis (H₁)

The average price is significantly different between phones with and without a warranty.

## 5. Conclusion

- There is a **statistically significant relationship between `Warranty` and `Price`**.
- Phones with a warranty have a **higher average and median price** compared to phones without a warranty.
- However, this result shows an **association**, not necessarily that a warranty directly causes a higher phone price.
- The large difference in group sizes should also be considered because there are many more phones without a warranty than with a warranty.
