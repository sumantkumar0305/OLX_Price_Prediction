# EDA on Phone_Name Column

## 1. Data Quality and Missing Values

- The `Phone_Name` column has **no missing values**.
- This means every row in the dataset has a phone name.
- There are **575 unique phone models** in the dataset.
- This means the column has **high cardinality**, as there are many different phone models.

## 2. Percentage Distribution

- **iPhone 13** is the most common phone model, making up around **3.68%** of the dataset.
- **iPhone 15** is the second most common model, making up around **3.55%**.
- Other popular models include **iPhone 11, iPhone 14, and iPhone 16**.
- Many other phone models appear only a few times, with some accounting for around **0.062%** of the dataset.
- This shows that a few phone models are very common, while many models have very low frequency.

## 3. Visual Analysis

- The **iPhone 13** appears nearly **59 times** in the dataset.
- The **iPhone 15** appears around **57 times**.
- Apple phones are clearly more common in the dataset.
- Among the **top 15 phone models, 14 are iPhones**.
- The only non-Apple phone in the top 15 is the **Samsung Galaxy S24 Ultra**, which appears fewer than 20 times.

## 4. Conclusion

- The `Phone_Name` column contains **575 different phone models**.
- There are **no missing values** in this column.
- The distribution is **highly uneven**.
- **Apple iPhones dominate** the dataset, while most other phone models appear only a few times.
- Therefore, `Phone_Name` is a **high-cardinality categorical feature**.
